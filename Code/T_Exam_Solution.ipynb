```python
# задача 2

def max_Boris_flowers(n):
    for _ in range(n):
        a = int(input())

        if a < 7:
          print(-1)
          continue


        def find_max_bouquet(a):
            for i in range(64, -1, -1):
                for j in range(i - 1, -1, -1):
                  for k in range(j - 1, -1, -1):

                    cost = (1 << i) + (1 << j) + (1 << k)
                    if cost <= a:
                      return cost
            return -1

        print(find_max_bouquet(a))

n = int(input())
max_Boris_flowers(n)
```

```python
# задача 3

def minimal_adjustments(n, m, A):
    A1, A2 = A[0], A[1]
    adjustments = []

    for i in range(2, n):
        Ai = A[i]
        if A1 <= Ai <= A2:
            adjustments.append(0)
        else:
            if Ai < A1:
                adjustments.append(A1 - Ai)
            else:
                adjustments.append(Ai - A2)

    adjustments.sort()
    return sum(adjustments[:m])

n, m = map(int, input().split())
A = list(map(int, input().split()))

print(minimal_adjustments(n, m, A))
```

```python
# задача 4

def minimal_operations(n, x, y, z, A):
    def calculate_ops(a, divisor):
        if a % divisor == 0:
            return 0
        else:
            return divisor - (a % divisor)

    min_ops = float('inf')

    for i in range(n):
        a = A[i]

        ops_x = calculate_ops(a, x)
        new_a = a + ops_x
        ops_y = calculate_ops(new_a, y)
        new_a += ops_y
        ops_z = calculate_ops(new_a, z)
        total_ops = ops_x + ops_y + ops_z
        if total_ops < min_ops:
            min_ops = total_ops

    for i in range(n):
        for j in range(n):
            if i == j:
                continue

            ops_x = calculate_ops(A[i], x)
            new_a_i = A[i] + ops_x
            ops_y = calculate_ops(new_a_i, y)
            new_a_i += ops_y
            ops_z = calculate_ops(A[j], z)
            total_ops = ops_x + ops_y + ops_z
            if total_ops < min_ops:
                min_ops = total_ops

            ops_x = calculate_ops(A[i], x)
            new_a_i = A[i] + ops_x
            ops_z = calculate_ops(new_a_i, z)
            new_a_i += ops_z
            ops_y = calculate_ops(A[j], y)
            total_ops = ops_x + ops_z + ops_y
            if total_ops < min_ops:
                min_ops = total_ops

            ops_y = calculate_ops(A[i], y)
            new_a_i = A[i] + ops_y
            ops_z = calculate_ops(new_a_i, z)
            new_a_i += ops_z
            ops_x = calculate_ops(A[j], x)
            total_ops = ops_y + ops_z + ops_x
            if total_ops < min_ops:
                min_ops = total_ops

    for i in range(n):
        for j in range(n):
            for k in range(n):
                if i == j or j == k or i == k:
                    continue

                ops_x = calculate_ops(A[i], x)
                ops_y = calculate_ops(A[j], y)
                ops_z = calculate_ops(A[k], z)
                total_ops = ops_x + ops_y + ops_z
                if total_ops < min_ops:
                    min_ops = total_ops

    return min_ops

n, x, y, z = map(int, input().split())
A = list(map(int, input().split()))

print(minimal_operations(n, x, y, z, A))
```

```python
# задача 5

def main(n, s, a):
    prefix = [0] * (n + 1)
    for i in range(n):
        prefix[i + 1] = prefix[i] + a[i]

    total = 0

    for l in range(1, n + 1):
        for r in range(l, n + 1):
            S = prefix[r] - prefix[l - 1]
            if S <= s:
                total += 1
            else:
                total += (S + s - 1) // s

    print(total)

n, s = map(int, input().split())
a = list(map(int, input().split()))

main(n, s, a)
```

```python
# задача 6

def max_happy_triplets(n, points):
    def is_happy_triangle(p1, p2, p3):
        (x1, y1), (x2, y2), (x3, y3) = p1, p2, p3
        area = (x2 - x1) * (y3 - y1) - (y2 - y1) * (x3 - x1)
        return area != 0

    used = [False] * n
    count = 0

    for i in range(n):
        if used[i]:
            continue
        for j in range(i + 1, n):
            if used[j]:
                continue
            for k in range(j + 1, n):
                if used[k]:
                    continue
                if is_happy_triangle(points[i], points[j], points[k]):
                    count += 1
                    used[i] = used[j] = used[k] = True
                    break
            if used[i]:
                break
        if used[i]:
            continue

    return count

n = int(input())
points = [tuple(map(int, input().split())) for _ in range(n)]

print(max_happy_triplets(n, points))
```
